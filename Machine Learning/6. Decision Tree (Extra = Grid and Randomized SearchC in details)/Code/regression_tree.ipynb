````python
# Every column is numeric.
df = pd.read_csv(r"D:\VS CODE\Data Science\Machine Learning\6. Decision Tree\Dataset\HousingData.csv")

X = df.iloc[:, :-1]
y = df.iloc[:,  -1]

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 2)

reg_tree = DecisionTreeRegressor(max_depth = 10)
reg_tree.fit(X = x_train, y = y_train)

print(f"For first value of x_test, prediction = {reg_tree.predict(X = x_test[:5])}.\n")
print(f"root_mean_squared_error = {root_mean_squared_error(y_true = y_test, y_pred = reg_tree.predict(x_test))}.")
print(f"r2_score = {r2_score(y_true = y_test, y_pred = reg_tree.predict(x_test))}.")

plt.figure(figsize = (10, 10))
plot_tree(decision_tree = reg_tree)

extra_info = """
Can't quantify accuracy because the output column is numeric.
"""
````
