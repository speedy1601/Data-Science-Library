````python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier

import numpy as np
import pandas as pd
````

#                                               Theory
```js
        In Machine Learning "Feature Selection" is very important because not every column contributes. Some/many columns are totally redundant for what we want to achieve and we don’t need them at all to train our model.

        One way to find out which INPUT FEATURES has how much 'importance' for the OUTPUT COLUMN is 'feature_importances_' which we can calculate through Decision Tree which uses one model.
        Random Forest uses multiple Decision Trees. So when each model returns the importance of INPUT FEATURES, random forest returns the AVERAGE IMPORTANCE of each column|feature.

        The summation of all the IMPORTANCE OF INPUT FEATURES is always 1, normalized.

        If you’re somehow eager to learn the little maths of Feature Importance which I am not, watch from 11:10 -> https://youtu.be/R47JAob1xBY?si=NO6ftptTUPFmnO9a&t=670.
```

#                                               Dataset import

````python
heart = pd.read_csv(r"D:\VS CODE\Data Science\Machine Learning\8. Random Forest\Dataset\heart.csv") .fillna(0)
print(heart.head(5).to_string())

X, y = heart.iloc[:, :-1], heart.iloc[:, -1]
# No need to find x_train, y_train, x_test, y_test since to find the Feature Importance we need the whole actual data.
````

#                                   Feature Importance using Decision Tree

````python
dt = DecisionTreeRegressor(random_state = 42) .fit(X = X, y = y)

print(f"Features Importance = {dt.feature_importances_}.\n")
print(f"Summation of all the Features Importance = {dt.feature_importances_.sum()}.\n")
print(f"""Column "{heart.columns[dt.feature_importances_.argmax()]}" has the most Importance.""")
````

#                                   Feature Importance using Random Forest

````python
rf = RandomForestClassifier(n_estimators = 3, random_state = 42) .fit(X = X, y = y) # We're using 3 decision trees.

print(f"Features Importance = {rf.feature_importances_}.\n")
print(f"Summation of all the Features Importance = {rf.feature_importances_.sum()}.\n")
print(f"""Column "{heart.columns[rf.feature_importances_.argmax()]}" has the most Importance.\n""")

print(f"{'-' * 130}")

print(f"First  model's Feature Importance = {rf.estimators_[0].feature_importances_}.") # Numpy array.
print(f"Second model's Feature Importance = {rf.estimators_[1].feature_importances_}.") # Numpy array.
print(f"Third  model's Feature Importance = {rf.estimators_[2].feature_importances_}.\n") # Numpy array.

print(f"The mean values : {(rf.estimators_[0].feature_importances_ + rf.estimators_[1].feature_importances_ + 
                            rf.estimators_[2].feature_importances_) / rf.estimators_.__len__()}.")

explanation = """
The mean values i.e. Important Features, is the same as rf.feature_importances_ returned in one go.
"""
````
