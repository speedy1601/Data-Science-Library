```python
arr = np.array([-0.6, 0.4, -0.6, 0.4, 0.4])
def ss(i):
    ssl = (arr[:i].sum() ** 1) / (0.24 * i)
    ssr = (arr[i:].sum() ** 1) / (0.24 * (5 - i))
    return ssl, ssr, ssl + ssr

for i in range(1, 5):
    print(ss(i))
```

```python
def prob(arr):
    return 1 / (1 + np.exp(-arr))
log_to_probability = np.vectorize(pyfunc = prob)

pred_y_1 = log_to_probability([0.29, 0.29, 0.29, 0.57, 0.57])
placed = np.array([0, 1, 0, 1, 1])

for v in [f"{v:.2f}" for v in placed - pred_y_1]:
    print(v)
```
