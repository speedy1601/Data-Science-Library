```python
import plotly.express as px
import pandas as pd
from dash import Dash, html, dcc, callback, Output, Input, dash_table

# Load the iris dataset into Polars
iris:pd.DataFrame = px.data.iris()

app = Dash()

app.layout = html.Div([ 
    html.H1(children = "Iris Data Table", style={'textAlign' : 'center'}),
    
    dcc.Dropdown(
        id = 'flower_name',
        options = iris['species'].unique(),
        value = 'setosa',
        placeholder = 'Select Flower Name',

        style = {'width': '700px',
                 'margin': 'auto',
                 'backgroundColor': "#18B1CC6C",
                 'color': "#000000A3",             
                 'fontFamily': "Segoe UI",
                 'border': '1px solid whitesmoke',
                 'fontSize' : '19px',}
    ),
    
    dash_table.DataTable(
        id='output_table',
        data=[],

        style_table={
            'height': '500px',
            'width': '700px',
            'overflowY': 'auto',
            'backgroundColor': '#000',   # pitch-black container
            'margin': 'auto'
        },
        style_header={
            'backgroundColor': "#000000DC",   # very dark  -- rgba(2, 158, 247, 0.5)
            'color': "#8A97AA",             # neon green
            'fontFamily': "Segoe UI",
            'fontWeight': 'semi-bold',
            'border': '1px solid whitesmoke',
            'fontSize' : '19px',
        },
        style_cell={
            'backgroundColor': "#1D1F22DC",
            'color': "#8A97AA",
            'fontFamily': "Segoe UI",
            'textAlign': 'center',
            'minWidth': '20px',
            'whiteSpace': 'normal',
            'fontSize' : '18px',
        },
        css=[
        {
            'selector': 'table',
            'rule': 'border-collapse: separate; border-spacing: 0 15px; margin-top: 5px;'
        },
        {
            'selector': 'tr:hover',
            'rule': 'background-color: #18B1CC6C;'
        }
    ],
        # style_data_conditional=[
        #     {
        #         'if': {'row_index': 'odd'},
        #         'backgroundColor': "#1c1d1d71"  # subtle striping
        #     }
        # ],
    )
])

@callback(
    Output(component_id = 'output_table', component_property = 'data'),
    Input(component_id = 'flower_name', component_property = 'value'),
)
def filter_table(flower_name): # flower_name is the 'value' of Dropdown.
    return ( iris.iloc[iris['species'].eq(flower_name).to_numpy(), :4]
            .to_dict(orient = 'records') ) # returns list[dict].

if __name__ == '__main__':
    # app.run(debug=True)
    print(1)
```

```python
import plotly.express as px
import pandas as pd
from dash import Dash, html, dcc, callback, Output, Input, dash_table

# Load the iris dataset into Polars
iris:pd.DataFrame = px.data.iris()

app = Dash()

app.layout = html.Div(
    style={'position' : 'relative'},

    children = [
        html.H1(children = "Iris Data Table", style={'textAlign' : 'center'}),

        html.Button(id = 'button1',
                    style = {'width': '200px', 'height': '100px', 'border': '2px solid red', 'color': 'blue'},
                    value = 'aaaaaaaaa'),

        dash_table.DataTable(
            id='output_table',
            data= iris.to_dict(orient = 'records'),

            style_table={
                'height': '500px',
                'width': '700px',
                'overflowY': 'auto',
                'backgroundColor': '#000',   # pitch-black container
                'margin': 'auto',
            },
            style_header={
                'backgroundColor': "#000000DC",   # very dark  -- rgba(2, 158, 247, 0.5)
                'color': "#8A97AA",
                'fontFamily': "Segoe UI",
                'fontWeight': 'semi-bold',
                'border': '1px solid whitesmoke',
                'fontSize' : '19px',
                'position' : 'sticky',
                'zIndex': 2,
                'top': 0, # style_header is child and the parent is the DataTable.
                
            },
            style_cell={
                'backgroundColor': "#1D1F22DC",
                'color': "#8A97AA",
                'fontFamily': "Segoe UI",
                'textAlign': 'center',
                'minWidth': '20px',
                'whiteSpace': 'normal',
                'fontSize' : '18px',
            },
            css=[
                {
                    'selector': 'table',
                    'rule': 'border-collapse: separate; border-spacing: 0 15px; margin-top: 5px;'
                },
                {
                    'selector': 'tr:hover',
                    'rule': 'background-color: #18B1CC6C;'
                }
            ],
        ),

        html.Br(),

        html.Div(id = 'floating_div',
                 children="I am floating!",
                 style = {'width': '700px',
                          'height': '250px',
                          'border': '2px solid greenyellow',
                          'margin': 'auto'}),

    ],
)


if __name__ == '__main__':
    # app.run(debug=True)
    print(1)
```
